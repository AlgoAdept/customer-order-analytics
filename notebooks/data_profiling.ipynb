```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
DATA_PATH = "../data/raw/SuperStoreOrders.csv"

df = pd.read_csv(DATA_PATH)

print(f"Dataset Shape : {df.shape}")

df.head()
```

```python
print("="*60)
print("DATASET INFORMATION")
print("="*60)

df.info()
```

```python
# BASIC STATS
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])

print("\nColumn Names\n")
print(df.columns.tolist())
```

```python
# MISSING VALUES 

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": round(df.isnull().mean()*100,2)
})

missing.sort_values("Missing Values", ascending=False)
```

```python
# DUPLICATE RECORDS

duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
# UNIQUE VALUES

unique = pd.DataFrame({
    "Unique Values": df.nunique()
})

unique
```

```python
# NUMERICAL STATS

numeric_summary = df.describe().transpose()
```

```python
# COVERTING DATE COLUMNS TO DATETIME FORMAT FOR ANALYSIS 

# Date columns

df["order_date"] = pd.to_datetime(df["order_date"], format='%d/%m/%Y', errors='coerce')

df["ship_date"] = pd.to_datetime(df["ship_date"], format='%d/%m/%Y', errors='coerce')

# Numeric columns

df["sales"] = pd.to_numeric(df["sales"], errors="coerce")
```

```python
df.dtypes
```

```python
# NUMERICAL SUMMARY

df.describe().T
```

```python
# CATEGORICAL SUMMARY 

categorical = df.select_dtypes(include="object")

categorical.describe().T
```

```python
# DATE ANALYSIS

print("Order Date Range")

print(df["order_date"].min())

print(df["order_date"].max())

print()

print("Ship Date Range")

print(df["ship_date"].min())

print(df["ship_date"].max())
```

```python
# CUSTOMER ANALYSIS

print("Unique Customers :", df["customer_name"].nunique())

print("Unique Orders :", df["order_id"].nunique())

print("Unique Products :", df["product_id"].nunique())
```

```python
# REGION DISTRIBUTION

df["region"].value_counts().to_frame("Customers")
```

```python
# MARKET DISTRIBUTION

df["market"].value_counts().to_frame("Orders")
```

```python
# SEGMENT DISTRIBUTION

df["segment"].value_counts().to_frame("Customers")
```

```python
# TOP CUSTOMERS BY ORDER COUNT

(
    df.groupby("customer_name")["order_id"]
      .count()
      .sort_values(ascending=False)
      .head(10)
      .to_frame("Orders")
)
```

```python
# TOP CUSTOMERS BY REVENUE

(
    df.groupby("customer_name")["sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .to_frame("Revenue")

)
```

```python
# SALES DISTRIBUTION

df["sales"].describe()
```

```python
# PROFIT ANALYSIS

print("Negative Profit Orders")

(df["profit"] < 0).sum()
```

```python
# DISCOUNT ANALYSIS

df["discount"].value_counts().sort_index()
```

```python
# SHIPPING ANALYSIS
df["ship_mode"].value_counts()
```

```python
# DATA QUALITY CHCEKS

print("="*60)
print("DATA QUALITY CHECKS")
print("="*60)

print("Null Values :", df.isnull().sum().sum())

print("Duplicate Rows :", df.duplicated().sum())

print("Negative Sales :", (df["sales"] < 0).sum())

print("Negative Profit :", (df["profit"] < 0).sum())

print("Future Dates :", (df["order_date"] > pd.Timestamp.today()).sum())
```

```python
# SUMMARY


summary = {
    "Total Orders": df["order_id"].nunique(),
    "Total Customers": df["customer_name"].nunique(),
    "Total Products": df["product_id"].nunique(),
    "Total Revenue": round(df["sales"].sum(),2),
    "Average Order Value": round(df["sales"].mean(),2),
    "Average Profit": round(df["profit"].mean(),2),
}

pd.DataFrame(summary.items(), columns=["Metric","Value"])
```
